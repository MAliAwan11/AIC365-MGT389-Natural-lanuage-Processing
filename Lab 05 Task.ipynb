### **Task 01**: Tokenization
### 1.	Write tokenize(sentence) that lowercases, splits on spaces and adds *"<s> and </s>"*.

```python
import math, itertools
from collections import Counter

train = ["the cat sat on the mat", "the cat ate the fish",
         "the dog sat on the rug", "the dog ate the bone"]
test  = ["the cat sat on the rug", "the fish sat on the mat"]

def tokenize(sentence):
    return ["<s>"] + sentence.lower().split() + ["</s>"]
```

```python
for s in train + test:
    print(tokenize(s))
```

```python
print("Tokens with markers   :", sum(len(tokenize(s)) for s in train))
print("Tokens without markers:", sum(len(s.split()) for s in train))
```

#### 4.	Write build_vocab(sentences) and print V and the sorted vocabulary.

```python
def build_vocab(sentences):
    words = {w for s in sentences for w in s.lower().split()}
    words.add("</s>")                     # V = all words plus </s> (no <s>)
    return sorted(words)

vocab = build_vocab(train)
V = len(vocab)
print("V =", V)
print(vocab)
```

#### 1.	Write get_ngrams(tokens, n) that returns a list of tuples.

```python
def get_ngrams(tokens, n):
    return [tuple(tokens[i:i + n]) for i in range(len(tokens) - n + 1)]
```

### **Task 04**: Unigram model
#### 1.	Write unigram_prob(w) using MLE, P(w) = C(w) / N.

```python
N = sum(c for (w,), c in uni.items() if w != "<s>")      
def unigram_prob(w):
    return uni.get((w,), 0) / N
```

#### 3.	Write unigram_sentence_prob(sentence) and print the result for both test sentences.

### **Task 05**: Bigram model (MLE)
#### 1.	Write bigram_prob(prev, w) using P(w | v) = C(v, w) / C(v).


```python
def bigram_prob(prev, w):
    c = uni.get((prev,), 0)
    return bi.get((prev, w), 0) / c if c else 0.0
```

#### 2.	Write next_word_distribution(prev) that returns a dictionary of next words and probabilities. Print it for “the” and assert that it sums to 1.

```python
def next_word_distribution(prev):
    c = uni.get((prev,), 0)
    return {b: bi[(a, b)] / c for (a, b) in bi if a == prev} if c else {}

d = next_word_distribution("the")
print("P(next | the):", {k: round(v, 4) for k, v in d.items()})
assert abs(sum(d.values()) - 1) < 1e-9
```

#### 3.	Write bigram_sentence_prob(sentence) using the chain rule and print the result for both test sentences.

```python
def bigram_sentence_prob(sentence):
    p = 1.0
    for prev, w in get_ngrams(tokenize(sentence), 2):
        p *= bigram_prob(prev, w)
    return p

for s in test:
    print(f"P({s!r}) = {bigram_sentence_prob(s):.4e}")
```

#### 4.	Write predict_next(prev) that returns the most probable next word (break ties alphabetically) and test it on at least five different words.

```python
def predict_next(prev):
    dist = next_word_distribution(prev)
    return min(dist, key=lambda w: (-dist[w], w)) if dist else None   

for w in ["<s>", "the", "cat", "dog", "sat", "ate", "on"]:
    print(f"predict_next({w!r}) = {predict_next(w)!r}")
```

### **Task 06**: Trigram model
#### 1.	Write trigram_prob(w1, w2, w3) using P(w3 | w1, w2) = C(w1, w2, w3) / C(w1, w2).

```python
def trigram_prob(w1, w2, w3):
    c = bi.get((w1, w2), 0)
    return tri.get((w1, w2, w3), 0) / c if c else 0.0
```

#### 2.	Write trigram_sentence_prob(sentence). Use the bigram model for the first word given <s> and the trigram model for all other words. Print the result for both test sentences.

```python
def trigram_sentence_prob(sentence):
    toks = tokenize(sentence)
    p = bigram_prob("<s>", toks[1])                  
    for w1, w2, w3 in get_ngrams(toks, 3):           
        p *= trigram_prob(w1, w2, w3)
    return p

for s in test:
    print(f"P({s!r}) = {trigram_sentence_prob(s):.4e}")
```

### **Task 08**: Smoothing
#### 1.	Write bigram_prob_addone(prev, w) using (C(v, w) + 1) / (C(v) + V).

```python
def bigram_prob_addone(prev, w):
    return (bi.get((prev, w), 0) + 1) / (uni.get((prev,), 0) + V)
```

#### 2.	Print P(sat | fish) and P(cat | the) under both MLE and add-one smoothing, side by side.

```python
print(f"{'':<16}{'MLE':>8}{'Add-one':>10}")
print(f"{'P(sat | fish)':<16}{bigram_prob('fish','sat'):>8.4f}{bigram_prob_addone('fish','sat'):>10.4f}")
print(f"{'P(cat | the)':<16}{bigram_prob('the','cat'):>8.4f}{bigram_prob_addone('the','cat'):>10.4f}")
```

#### 4.	Write bigram_prob_addk(prev, w, k) and print P(sat | fish) for k = 1, 0.5, 0.1 and 0.01 in a table

```python
def bigram_prob_addk(prev, w, k, uni_c=None, bi_c=None, V_=None):
    uni_c = uni if uni_c is None else uni_c
    bi_c  = bi  if bi_c  is None else bi_c
    V_    = V   if V_    is None else V_
    return (bi_c.get((prev, w), 0) + k) / (uni_c.get((prev,), 0) + k * V_)

print(f"{'k':<8}{'P(sat | fish)':>14}")
for k in [1, 0.5, 0.1, 0.01]:
    print(f"{k:<8}{bigram_prob_addk('fish','sat',k):>14.4f}")
```

### **Task 09**: Perplexity and model comparison
#### 1.	Write perplexity(sentences, prob_fn) that returns infinity when any probability is 0.

```python
def perplexity(sentences, prob_fn):
    log_sum, count = 0.0, 0
    for s in sentences:
        for prev, w in get_ngrams(tokenize(s), 2):
            p = prob_fn(prev, w)
            if p == 0:
                return float("inf")
            log_sum += math.log(p); count += 1
    return math.exp(-log_sum / count)
```

#### 2.	Compute the perplexity of the two test sentences together for the bigram MLE model, the bigram add-one model and the unigram add-one model, where P(w) = (C(w) + 1) / (N + V). Print one results table.

```python
unigram_addone = lambda prev, w: (uni.get((w,), 0) + 1) / (N + V)
print(f"{'Model':<24}{'Perplexity':>12}")
print(f"{'Bigram MLE':<24}{perplexity(test, bigram_prob):>12.4f}")
print(f"{'Bigram add-one':<24}{perplexity(test, bigram_prob_addone):>12.4f}")
print(f"{'Unigram add-one':<24}{perplexity(test, unigram_addone):>12.4f}")
```

#### 3.	Compute the perplexity of the add-k bigram model for k = 1, 0.5, 0.1 and 0.01. Print the table and the best k.

```python
ks = [1, 0.5, 0.1, 0.01]
pp_k = {k: perplexity(test, lambda a, b, k=k: bigram_prob_addk(a, b, k)) for k in ks}
print(f"{'k':<8}{'Perplexity':>12}")
for k in ks:
    print(f"{k:<8}{pp_k[k]:>12.4f}")
best_k = min(pp_k, key=pp_k.get)
print("Best k on the test set (full training data):", best_k)
```

#### 4.	Train only on the first three training sentences (build V from all four) and pick the best k using the fourth sentence as validation data. Report the test perplexity with that k, and compare it with the best k found directly on the test set.

```python
def build_counts(sentences):
    u, b = Counter(), Counter()
    for s in sentences:
        t = tokenize(s)
        u.update(get_ngrams(t, 1)); b.update(get_ngrams(t, 2))
    return u, b

u3, b3 = build_counts(train[:3])             # train on the first three sentences only
V_all = len(build_vocab(train))              # but build V from all four
val = [train[3]]                             # fourth sentence = validation data

fn = lambda k: (lambda a, b: bigram_prob_addk(a, b, k, u3, b3, V_all))
pp_val   = {k: perplexity(val,  fn(k)) for k in ks}
pp_test3 = {k: perplexity(test, fn(k)) for k in ks}

print(f"{'k':<8}{'Validation PP':>15}{'Test PP':>12}")
for k in ks:
    print(f"{k:<8}{pp_val[k]:>15.4f}{pp_test3[k]:>12.4f}")
k_val  = min(pp_val,   key=pp_val.get)
k_test = min(pp_test3, key=pp_test3.get)
print("Best k on validation:", k_val,  "-> test perplexity:", round(pp_test3[k_val], 4))
print("Best k on test      :", k_test, "-> test perplexity:", round(pp_test3[k_test], 4))
```
