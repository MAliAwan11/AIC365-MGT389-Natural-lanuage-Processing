## Task 2: Pattern-Based Extraction (Real World Data)

### Given a paragraph containing emails, phone numbers, order IDs, dates, and percentages mixed together:
#### a) Write a regex pattern using re.findall() to extract all email addresses
#### b) Write a regex pattern to extract all phone numbers (assume format: 03XX-XXXXXXX)
#### c) Write a regex pattern to extract all dates in DD/MM/YYYY format
#### d) Write a regex pattern to extract all percentage values (e.g., 45%, 12.5%)


```python
paragraph = """
EMAIL: muhammadali@mail.com
Alternate EMAIL: maliawan@mail.com

NUMBER: 0310-1234567
Alternate NUMBER: 0300-1234567

Order 1 ID: ORD-001
Date: 29/09/2026
Completion status: 100%

Order 2 ID: ORD-002
Date: 30/09/2026
Completion status: 75%
"""
```

```python
emails = re.findall(r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b', paragraph)
print("a) Emails:", emails)
```

```python
phones = re.findall(r'\b03\d{2}-\d{7}\b', paragraph)
print("b) Phone numbers:", phones)
```

```python
dates = re.findall(r'\b\d{2}/\d{2}/\d{4}\b', paragraph)
print("c) Dates:", dates)
```

```python
order_ids = re.findall(r'\bORD-\d{3}\b', paragraph)
print("d) Order IDs:", order_ids)
```

```python
percentages = re.findall(r'\b\d+(?:\.\d+)?%', paragraph)
print("e) Percentages:", percentages)
```

## Task 4: Token Attributes (Categorizing Tokens)

### Using a sample sentence that contains normal words, numbers, an email address, and a currency symbol (e.g., "$50" or "Rs. 500"):
#### a) Load the sentence into a spaCy Doc object
#### b) Loop through each token and print its is_alpha value
#### c) Print each token's like_num and like_email values
#### d) Identify and list only the tokens where is_currency is True


```python
sample = "Hi there, I am Ali, Roll Number: SP23-BBD-030, Email: maliawan108@gmail.com and Phone Number: 0310-5671517, and my product is for $5.00 "
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")
doc = nlp(sample)
```

```python
print("is_alpha (True/False)")
for token in doc:
    print(token.text, token.is_alpha)
```

```python
print("like_num (True/False) & like_email (True/False)")
for token in doc:
    print(token.text, token.like_num, token.like_email)
```

```python
print("\n=== Currency Tokens ===")
for token in doc:
    if token.is_currency:
        print(token.text)
```

## Task 5: Sentence Segmentation (Handling Tricky Cases)

### Using a paragraph that includes at least one abbreviation (e.g., "Dr.", "Mr.", "e.g.") in the middle of a sentence:
#### a) Load a trained spaCy pipeline (en_core_web_sm)
#### b) Use doc.sents to split the paragraph into individual sentences
#### c) Print the total number of sentences detected
#### d) Check whether the abbreviation caused any incorrect sentence break, and explain your observation in one line



```python
paragraph = """
Dear Sir,
I hope this finds you well.
This is a simulation where I will pretend to be a doctor.
Regards,
Dr. Ali
Date: 30/09/2026
email: maliawan108@gmail.com
"""
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")
doc = nlp(paragraph)
```

```python
print("Detected Sentences:")
for i, sent in enumerate(doc.sents, 1):
    print(f"{i}. {sent.text.strip()}")
```

```python
print("\nTotal sentences detected:", len(list(doc.sents)))
```

##### Observation: No incorrect breaks — spaCy's trained model correctly treated "Dr." as an abbreviation, not a sentence ending.
